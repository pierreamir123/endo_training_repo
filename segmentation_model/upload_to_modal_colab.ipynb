```
!modal volume create prad-data
# trailing "/." puts image/, label/ at the Volume root (what modal_train.py expects at /data)
!modal volume put prad-data "{root}/image" /image
!modal volume put prad-data "{root}/label" /label
!modal volume put prad-data "{root}/label distribution.txt" "/label distribution.txt"
!modal volume ls prad-data /
```
